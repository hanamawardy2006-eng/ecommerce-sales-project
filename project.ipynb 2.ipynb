```python
import pandas as pd
import numpy as np
import os
```

```python


print("==================================================")
print("1. DATASET UNDERSTANDING & LOADING")
print("==================================================")

# File detection
file_name = 'ecommerce_sales_34500_2.csv'
if not os.path.exists(file_name):
    target = [f for f in os.listdir('.') if f.startswith('ecommerce') and f.endswith('.csv')]
    file_name = target[0] if target else 'ecommerce_sales_34500.csv'

df = pd.read_csv(file_name)
print(f"Dataset File Loaded: '{file_name}'")
print(f"Raw Dataset Shape: {df.shape[0]} Rows, {df.shape[1]} Columns")

print("\n==================================================")
print("2. DATA INSPECTION (MISSING, DUPLICATES, TYPES, OUTLIERS)")
print("==================================================")

# A. Data Types Check
print("\n--- Data Types ---")
print(df.dtypes)

# B. Missing Values Check
print("\n--- Missing Values Check ---")
print(df.isnull().sum())

# C. Duplicates Check
duplicates_count = df.duplicated().sum()
print(f"\n--- Duplicates Count: {duplicates_count} ---")

# D. Outliers Inspection (IQR Method)
print("\n--- Outliers Detection Summary (IQR Method) ---")
numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = ((df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR))).sum()
    print(f"Column '{col}': {outliers} potential outliers detected (Q1={Q1:.2f}, Q3={Q3:.2f})")

print("\n==================================================")
print("3. DATA CLEANING & FEATURE ENGINEERING")
print("==================================================")

# Deduplication & Datetime conversion
df_cleaned = df.drop_duplicates().copy()
df_cleaned['order_date'] = pd.to_datetime(df_cleaned['order_date'])

# Calculated Columns (Feature Engineering)
# 1. Gross Amount: Total sales value before applying discount
df_cleaned['gross_amount'] = (df_cleaned['price'] * df_cleaned['quantity']).round(2)

# 2. Discount Amount: Exact dollar value discounted per order
df_cleaned['discount_amount'] = (df_cleaned['gross_amount'] - df_cleaned['total_amount']).round(2)

# 3. Estimated Profit: Monetary profit calculated from profit margin percentage
df_cleaned['estimated_profit'] = (df_cleaned['total_amount'] * (df_cleaned['profit_margin'] / 100)).round(2)

# 4. Discount Tier: Categorizing discount intensity
df_cleaned['discount_tier'] = pd.cut(
    df_cleaned['discount'], 
    bins=[-0.01, 0.0, 0.10, 0.20, 1.0], 
    labels=['No Discount', 'Low (1-10%)', 'Medium (11-20%)', 'High (>20%)']
)

# 5. Customer Age Group: Categorizing demographic age brackets
df_cleaned['age_group'] = pd.cut(
    df_cleaned['customer_age'], 
    bins=[0, 25, 40, 60, 120], 
    labels=['Gen Z (<=25)', 'Millennials (26-40)', 'Gen X (41-60)', 'Seniors (>60)']
)

# 6. Temporal Features: Year, Month Name, Day Name
df_cleaned['order_year'] = df_cleaned['order_date'].dt.year
df_cleaned['order_month_name'] = df_cleaned['order_date'].dt.strftime('%B')
df_cleaned['order_day_name'] = df_cleaned['order_date'].dt.strftime('%A')

print("✅ Data Cleaning and Feature Engineering completed successfully!")

print("\n==================================================")
print("4. EXPORT FINAL CLEANED DATASET FOR POWER BI")
print("==================================================")

output_filename = 'Cleaned_Ecommerce_Sales.csv'
df_cleaned.to_csv(output_filename, index=False)
print(f"Cleaned dataset saved as '{output_filename}' ({df_cleaned.shape[0]} rows, {df_cleaned.shape[1]} columns)")

# Preview top 5 rows
df_cleaned.head()
```
